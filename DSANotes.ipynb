# **Backtracing**
1. for loop backtracking recursion tree. 
    -> It gives all subset of a given set. At every call set is a valid subset.(Subsequence)  
    Tri:
```
    []
    ├── [1]
    │   ├── [1,2]
    │   │   └── [1,2,3]
    │   └── [1,3]
    ├── [2]
    │   └── [2,3]
    └── [3]
```


````python
# Code :
res = []
nums = [1,2,3]
def backtrack(start, path):
        res.append(path.copy())
        for i in range(start, len(nums)):
            path.append(nums[i])   # choose
            backtrack(i+1, path)   # explore
            path.pop()             # undo
backtrack(0,[])
print(*res,sep='\n')
````

2. Subsequences using recursion [Pick , Not pick]
3. Recursion tree
```
        left -> Pick
        right -> Not pick

        lvl->0                     [],0
                                /       \
                            /         \
        lvl->1             [1],1         [],1
                            /  \          / \ 
                        /     \        /   \
        lvl->2          [1,2],2  [1],2  [2],2  [],2

        return Only when idx == len(nums)
        final return res = [[1,2],[1],[2],[]]
```
4. **If you need to return only one result, make the function return boolean. Append the path to res whene true basecase.**
5. **If you need to return count of results, make the function return integer. In function retrun left+right func.**

````python
# Subsequent arrays
def printF(res,path,idx,nums):
    if idx == len(nums):
        return res.append(path.copy())
    #Pick
    printF(res,path+[nums[idx]],idx+1,nums)
    #Not pick
    printF(res,path,idx+1,nums)
    return res

nums = [1,2,3]
res = printF([],[],0,nums)
print(*res , sep='\n')
````

# Merge sort
1. Understanding mergsort gives better grip on recursion and backtracking.
2. And divide and conquer too.
3. Try to write code without slicing the array. Just by using pointers.

````python
#merge sort
def mergeSort(nums):
    def merge(left,right):
        res = []
        i = 0
        j = 0
        while i<len(left) and j< len(right):
            if left[i] < right[j]:
                res.append(left[i])
                i +=1
            else:
                res.append(right[j])
                j+=1
        # Check for elements in left subarray
        while i<len(left):
            res.append(left[i])
            i+=1
        # check for right subarray
        while j<len(right):
            res.append(right[j])
            j+=1
        return res

    def divide(arr):
        if len(arr) <= 1:
            return arr
        mid = len(arr) // 2
        left = divide(arr[:mid])
        right= divide(arr[mid:])
        return merge(left,right)

    return divide(nums)

print(mergeSort([2,1,4,3,5,8,6]))
            
        
````

````python
#Quick sorting
def quickSort(nums,l,r):
    i = l
    j = r
    pivot = l
    while i<j:
        if nums[pivot] < nums[i]:
            while j>i:
                if nums[j] < nums[pivot]:
                    nums[i],nums[j] = nums[j],nums[i]
                j-=1
        i +=1
    nums[i],nums[pivot] = nums[pivot],nums[i]
    return nums

print(quickSort([2,1,4,3,5,8,6],0,6))
````

### BFS and DFS
> Use deque for BFS(Level order travers)   
> Use STACK if needed. ELSE use recurrsive approach

````python
# BFS or LEVEL ORDER TRAVERS
from collections import deque
def bfs(root):
    q = deque([root])
    res = []
    while(q):
        path = []
        for _ in range(len(q)):
            node = q.popleft()
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
            path.append(node.val)
        res.append(path)

# DFS => INORDER, PREORDER, POSTORDER
# Recurrsive method
res = []
def dfs(root):
    if not root:
        return
    dfs(root.left)          # Inorder
    res.append(root.val)
    dfs(root.right)

# DFS using STACK
def dfs_iterative(root):
    if not root:
        return
    stack = [root]
    while stack:
        node = stack.pop()
        print(node.val)
        # Push right first so left is processed first (LIFO)
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)   
````

## heapq
> Use Heap tree if insertion is continuous and need to find kth max or kth min element at each insertion.
> * For each subtree. Root element is smallest than left and right. And left is smaller than right.  
> * It is a _**priority queue**_.
> * In min_heap -> smallest element will be at zeroth index. Smallest element will be having high priority
> * In max_heap -> Largest element will be at zeroth index. Largest element will be having high priority.

1. Python **heapq** provides min heap.
2. This class will modify the list inplace.

- Functions of heapq
    - heapq.heapify(list_of_nums)
    - heapq.heappop(list_of_nums)
    - heapq.heappush(list_of_nums,val)
