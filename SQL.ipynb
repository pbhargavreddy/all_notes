# QUERY
1. **SELECT** col_name_1,col_name_2,.. FROM table_name;  -> Can use * for selecting all columns
2. **WHERE**.   -> Can use **AND, OR, NOT, BETWEEN, In** 
```
    SELECT *
    FROM Students
    WHERE branch IN ('AI','ECE');
```
3. **LIKE**     -> '%' Zero or more chars. -> '_' Exactly on char
```
    SELECT *
    FROM Students
    WHERE name LIKE 'R%';    -> End with r => '%R'
```
4. **DISTINCT** -> Used before col names in select query. 
5. **AS**   -> Display col_name insted of.
6. **ORDER BY** -> Sort in ascending order.
```
    SELECT *
    FROM Students
    ORDER BY cgpa;      -> ORDER BY cgpa DESC; will sort in descending order.
```
7. **LIMIT** -> Returns only specified number of rows. Written at last line of query.
```
    SELECT *
    FROM Students
    ORDER BY cgpa DESC
    LIMIT 3;
```
8. **HAVING** -> Filters after grouping. **WHERE** filters befor grouping

## Execution order
```
    SELECT name
    FROM Students
    WHERE cgpa > 8
    ORDER BY cgpa DESC
    LIMIT 2;
```
___

```
SELECT branch,
       AVG(cgpa)
FROM Students
GROUP BY branch
HAVING AVG(cgpa) > 8.5;
```
