## I. INHERITANCE
1. It helps a child class to inherit parent's methods and constructors.

```python
# Super()
class Animal():
    def __init__(self,name):
        self.name  = name

    def info(self):
        print("Animal Name: ",self.name)

class Dogesh(Animal):
    def __init__(self,name,breed):
        super().__init__(name)     #Watch out. -> super() will help child class to inherit variables or attributes or methodfrom parent class.
        self.breed = breed
    def details(self):
        print("Name: ",self.name,"\nBreed: ",self.breed)
    

# Try multiple inheritance(inheriting multiple classes at a time.)
# Try multilevel inheritance

if __name__ == "__main__":
    dog = Dogesh("Buddy","goldern retriver")
    dog.info()
    dog.details()
```

```python
# Method overwritting
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

d = Dog()
d.sound()
```

## II. ENCAPSULATION
1. <u>**_Bunduling data attributes and methods into a single unit(class) and restricting direct access to to the data._**</u>
2. Using self.__variable in constructor, python name **_mangaling_** machanism will make it not accessible to child classes.

* It keeps code private. By declaring a member private, it is only accessable and modifiable by same class. 

```python
class BankAccount:

    def __init__(self):
        self.__balance = 10000
        self.name = "Bhargav"

    # update
    def deposit(self, amount):
        self.__balance += amount

    # getter function
    def get_balance(self):
        return self.__balance
    # setter function
    def set_balance(self,value):
        self.__balance = value

    # This function can not be used outside the class. 
    def __privateFunc(self):
        return "inside private function"        

b1 = BankAccount()    
# print(b1.__balance)     #-> gives error. Because it is private var.
b1.__balance = 2000
print("Local __balance: ",b1.__balance)       #-> This is going to print 2000. Because in __main__ b1.__balance = 2000

print("Class __balance",b1.get_balance())     #-> This is a getter function
b1.set_balance(12000)                         #-> This is a setter.
print("Class __balance",b1.get_balance())     #-> This is a getter function

# print(b1.__privateFunc())                     #-> Pribate method. won't work outside class.
print()

# Try with inheritance
class Scam(BankAccount):

    def __init__(self):
        # Try to inherit __balance
        super().__init__()

    def info(self):
        print(self.__balance)       # Will give error.

    
s = Scam()
# s.info()    # Will give the error. says Scam class doesn't have attribute  "__Scam__balance" 
print("Inherited value:",s.get_balance()) # -> This will work. Beacause the getter function is defined in Parent class.
```

## III. ABSTRACTION
1. <u>_**Showing only essential feature while hiding implementation details.**_</u>
2. Abstraction hides implementation details and exposes only the required functionality.
3. However **inspect** module will help to have a peek into the implementation.

```python
from abc import ABC, abstractmethod

# 1. Define the Abstract Class (Blueprint)
class PaymentProcessor(ABC):
    
    @abstractmethod
    def pay(self, amount: float):
        """Processes the payment. Must be implemented by concrete classes."""
        pass

    @abstractmethod
    def refund(self, amount: float):
        """Processes a refund."""
        pass


# 2. Concrete Implementation 1: Credit Card
class CreditCardPayment(PaymentProcessor):
    def pay(self, amount: float):
        print(f"Paid ${amount} via Credit Card.")
        
    def refund(self, amount: float):
        print(f"Refunded ${amount} to Credit Card.")


# 3. Concrete Implementation 2: PayPal
class PayPalPayment(PaymentProcessor):
    def pay(self, amount: float):
        print(f"Paid ${amount} via PayPal account.")
        
    def refund(self, amount: float):
        print(f"Refunded ${amount} via PayPal account.")


# --- Usage ---

# This will throw an ERROR! You cannot instantiate an abstract class directly:
# payment = PaymentProcessor()  # TypeError: Can't instantiate abstract class PaymentProcessor...

def process_order(processor: PaymentProcessor, amount: float):
    # The client code doesn't care HOW payment is processed,
    # only that it has a .pay() method available.
    processor.pay(amount)

cc = CreditCardPayment()
paypal = PayPalPayment()

process_order(cc, 100.0)      # Output: Paid $100.0 via Credit Card.
process_order(paypal, 50.0)   # Output: Paid $50.0 via PayPal account.

# PaymentProcess class is defined by the devoloper who is making e-commers website.
# ""He defined the Blue print of the payment. It is like ""contract"".""
""" PayPalPayment class and CreditCardPayment class are available options. 
Those availabe options must have the implementation of pay and refund methods.

Here where is link for hiding complex implementation. 
The payment gatways doesn't need to know implementation of Process_order. Instead the  devoluper make an abstract class.
Which payment gatway providers used.

And the devoluper don't need to know the implementation of 
"""
```

## IV. POLYMORPHISM
1. Method overriding -> **Run-time polymorphism. Method is confirmed at run time**
2. Method overloading -> **Compile-time polymorphism. Methos is confirmed at compile time.**

```python
""" Method Overloading
    Python doesn't support Overloading like Java cpp. Instead it will Achive it by using *args, **kwargs
"""

class Calculator:
    def multiply(self, a=1, b=1, *args):
        result = a * b
        for num in args:
            result *= num
        return result

# Create object
calc = Calculator()

# Using default arguments
print(calc.multiply())            
print(calc.multiply(4))           

# Using multiple arguments
print(calc.multiply(2, 3))       
print(calc.multiply(2, 3, 4))
```

```python
#Method Override
class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

# Polymorphic behavior
animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.sound())
```

### Public, private, protected  
Unlike java and c++, Python uses **_naming convention and name mangling_** not compiler restrictions. 
1. PUBLIC       -> self.name       --> Intented for everyone.
2. PROTECTED    -> self._name      --> Intended for parent and child class.
3. PRIVATE      -> self.__name     --> Intended for only class itself. 

> **(IMP)** Private members can not be used outside the class. Not even read.

## Static Variables and Instance Variables
* **Static or class variables.**  
-> Global  
->Static member belongs to class itself rather than to any specific instance of that class.  
-> They are same for every class. Declared globally in a class  
* **Instance variables.**  
-> declared in constructor.

```python
class Student:
    # Static variable. 
    college = "ABC Engineering College"  # Static variable or class variable


    # instance variable. They are specific for each instance
    def __init__(self,val):
        self.value = val

s1 = Student(20)
s2 = Student(47)


# Access static variables by using object or class
print(Student.college) 
print(s1.college,s1.value)
print(s2.college,s2.value)
```

# Static Methods and class methods And Instance method
* **Static method**  
-> It doesn't need acess to both class and instance variables. 
* **Class method**  
-> It doesn't need acess to instance variables. Needs **__cls__**
* **Instance method**  
-> Needs instance varibales.


```python
# Static method
class Math:


    # the following function doesn't work with class variables or instance variables. 
    # Used for helper functions. Or to give info of the class.
    # Static members can be inherited and overridden.
    @staticmethod
    def add(a, b):
        return a + b

# we can use the static method just for class
# m = Math()
# print(m.add(1,2))
print(Math.add(1,2))
```

```python
# class method
class Student:
    college = "ABC"


    # Here the following method works with class variables. Hence use @classmethod
    @classmethod
    def get_college(cls):
        return cls.college


# s = Student()
# print(s.get_college())
print(Student.get_college())
```

## Association, Aggrigation, Composition
1. Association is a relationship where two independent objects interact with each other.
2. Aggregation is a weak has-a relationship where the contained object can exist independently of the container.
3. Composition is a strong has-a relationship where the contained object depends on the lifetime of the container.


```python
# Association
class Teacher:
    def teach(self):
        print("Teaching")

class Student:
    def learn(self, teacher):
        teacher.teach()
```

```python
# Aggrigation
class Professor:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, professor):
        self.professor = professor

# Use
p = Professor("Dr. Rao")
d = Department(p)
```

```python
# Composition
class Engine:
    def start(self):
        print("Engine Started")

class Car:
    def __init__(self):
        self.engine = Engine()
```
