## I. INHERITANCE
1. It helps a child class to inherit parent's methods and constructors.

```python
# Super()
class Animal():
    def __init__(self,name):
        self.name  = name

    def info(self):
        print("Animal Name: ",self.name)

class Dogesh(Animal):
    def __init__(self,name,breed):
        super().__init__(name)           #Watch out. -> super() will help child class to inherit variables or attributes or methodfrom parent class.
        self.breed = breed
    def details(self):
        print("Name: ",self.name,"\nBreed: ",self.breed)
    

if __name__ == "__main__":
    dog = Dogesh("Buddy","goldern retriver")
    dog.info()
    dog.details()
```

```python
# Method overwritting
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

d = Dog()
d.sound()
```

## II. ENCAPSULATION
1. <u>**_Bunduling data attributes and methods into a single unit(class) and restricting direct access to to the data._**</u>
2. Using self.__variable in constructor, python name **_mangaling_** will machanism will make it not accessible to child classes.

```python
class BankAccount:

    def __init__(self):
        self.__balance = 10000

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

class Scam(BankAccount):

    def __init__(self):
        # Try to inherit __balance
        super().__init__()

    def info(self):
        print(self.__balance)       # Will give error.

s = Scam()
s.info()    # Will give the error. says Scam class doesn't have attribute  "__Scam__balance" 
```

## III. ABSTRACTION
1. <u>_**Showing only essential feature while hiding implementation details.**_</u>
2. Abstraction hides implementation details and exposes only the required functionality.
3. However **inspect** module will help to have a peek into the implementation.

```python
from collections import deque

class Animal(deque):
    pass
    # we don't need to know what is the implementation inside class deque. But still we can use it.

class Dog(Animal):
    pass
    # Mulilevel inheritance
```

## IV. POLYMORPHISM
1. Method overriding        # 
2. Method overloading

```python
""" Method Overloading
    Python doesn't support Overloading like Java cpp. Instead it will Achive it by using *args, **kwargs
"""

class Calculator:
    def multiply(self, a=1, b=1, *args):
        result = a * b
        for num in args:
            result *= num
        return result

# Create object
calc = Calculator()

# Using default arguments
print(calc.multiply())            
print(calc.multiply(4))           

# Using multiple arguments
print(calc.multiply(2, 3))       
print(calc.multiply(2, 3, 4))
```

```python
#Method Override
class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

# Polymorphic behavior
animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.sound())
```

# Static Variables, Class Variables, and Static Methods
1. Static means global. Every object have them same
2. 

```python
# Static variable
class Student:
    college = "ABC Engineering College"  # Static variable or class variable

s1 = Student()
s2 = Student()

print(s1.college)
print(s2.college)
```

```python
# Static method
class Math:

    @staticmethod
    def add(a, b):
        return a + b
```

```python
# class method
class Student:
    college = "ABC"

    @classmethod
    def get_college(cls):
        return cls.college
s = Student()
print(s.get_college())
```

## Association, Aggrigation, Composition
1. Association is a relationship where two independent objects interact with each other.
2. Aggregation is a weak has-a relationship where the contained object can exist independently of the container.
3. Composition is a strong has-a relationship where the contained object depends on the lifetime of the container.


```python
# Association
class Teacher:
    def teach(self):
        print("Teaching")

class Student:
    def learn(self, teacher):
        teacher.teach()
```

```python
# Aggrigation
class Professor:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, professor):
        self.professor = professor

# Use
p = Professor("Dr. Rao")
d = Department(p)
```

```python
# Composition
class Engine:
    def start(self):
        print("Engine Started")

class Car:
    def __init__(self):
        self.engine = Engine()
```
