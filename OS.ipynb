# Operating system
* An Operating System is system software that acts as an interface between the user/applications and computer hardware.  
* Eg: Windows, Linux, macOS, Android

___
## Responsibilities of os
```
Process Management
Memory Management
File Management
Device Management
Security
CPU Scheduling
Resource Allocation
```

___
1. **Kernel**  
-> The kernel is a <u>core part</u> of OS. It communicates with hardware directly.  
-> Applications cannot directly access hardware.They request services from the kernel using system calls.
2. **User mode**  
-> Applications run here. Example: Browser, VScode
-> Can not acces hardware directly. Can not execute privilaged programs. 
3. **Kernel mode**  
-> Operating system runs here.  
-> User mode is restricted for security, whereas kernel mode has complete access to hardware and system resources.
4. **Process**  
-> Process is a program in execution. Each has PID, Memory, Registers, Stack, Heap.  
-> While program is static, Stored on disk and have no resources allocated.

### States of process
```
        New → created

        Ready → waiting for CPU

        Running → executing

        Waiting → waiting for I/O

        Ready → back after I/O

        Terminated → finished
```

5. **Thread**  
-> A thread is a smallest execution of a process inside CPU. One process will have multiple threads.
6. **MultiThreading**  
-> Running multiple threads simultaniously. 
7. **Context switching**  
-> CPU switches from one process to another process. TO save resources and time.

___
## CPU Scheduling
* OS decides which process gets CPU. Now there are multi core cpu's every where. Scheduling is complex.
1. **FCFS** (First some first serve)
2. **SJF** (Shortest job first)
3. **Round Robin** => Each process gets <u>time Quantum</u>
4. **Priority scheduling**

___
### Deadlock
* It occurs whene two or more process wait forever for resources held by each other.
* Example:  

| Process | Holding | Required |
| ------- | ------- | -------- |
| A       | R1,R2   | R4       |
| B       | R3,R4   | R1       |  
* Then deadlock occurs.
* Conditions.
```
        Mutual Exclusion
        Hold and Wait
        No Preemption
        Circular Wait
```
* Break any one condition.


___
### Starvation vs Deadlock.
* Starvation -> Only one process waits forever because higher-priority processes keep getting CPU.
* Deadlock   -> No progress for any process.


___
### Memory management
* RAM allocation
1. **Pagging** removes **External fragmentation**  
-> Divides memory into equal space pages.  
-> Each page memory is divided into <u>Frames</u>  
2. **Fragmentation**   
-> **External fragmentation** means extra space in memory left between pages. Pagging will not leave space between pages.
-> **Internal fragmentation** means extra space left in each page. i.e. Allocation block is bigger than needed.
3. **Virtual memory**  
-> Allows programs larger than RAM to execute.  
-> Uses disk space as extension to RAM.
4. **Cache vs RAM**  
-> Cache is small and fast. It stores pre computation. So that no need to recompute all over again.
