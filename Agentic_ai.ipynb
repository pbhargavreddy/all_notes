## Agent
>-> A LLM will have information in global context. But it doesn't have permission to user environment.  
-> User create tools and let them interact with LLM. LLM will make the decision based on the instructions given by user.  
```
        Rule-Based AI
            ↓
        Machine Learning
            ↓
        Deep Learning
            ↓
        Generative AI
            ↓
        Large Language Models
            ↓
        Agentic AI
```

___
* An AI Agent is an AI system that can autonomously perceive information, reason about a goal, plan actions, use tools, remember relevant information, and execute tasks to achieve an objective.
1. **Tool calling**  
-> An LLM can not directly interact with user environment. Like sending e-mail, accessing database, Booking flight, calling API's
2. **Function calling**  
-> This is how LLM calls tools. User defines functions with **docstring description** at first line of function. LLM calls help() on the function and decides which to call and what parameters to pass.
3. **Memory**  
-> Instead of storing every previous response and prompt, User maintains short-term memory or long-term meomory.
4. **Planning**  
-> Break larger goal into smaller goals and performe each task.
5. **Reflection**  
-> Evaluate itself.This iterative improvement is often called reflection or self-correction.

___
6. **Multiagent system**  
-> Sometimes one agent isn't enough. So user need to devolup agents for subtasks.
7. **MCP(Model Context Protocol)**  
-> MCP provides a standardized way for AI models to communicate with external tools and data sources.  
-> Think of it like a USB. USB standardize communication.
-> Similarly MCP aims to standardize how AI applications interact with tools and services.
