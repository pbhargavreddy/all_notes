1. **Artificial neuron:** Recieves input from all neurons of previous layer. Calculates it's value updates it weights and bias.Each input has it's own weight.  
2. <u>_**Output of each neuron</u> = Activation(w1*x1 + w2*x2 + ... + b)**_
3. **Bias** allows the model to shift its decision instead of always starting from zero.


4. **Activation Function** : Without an activation function, a neural network is just a series of linear equations and cannot learn complex relationships.  
-> It introduces non linearity. Used for non-linear problems.  

| Function Name | Function Formula | Use Case |
| :--- | :--- | :--- |
| **Linear** | $f(x) = x$ | Output layer for continuous value regression. |
| **Sigmoid** | $f(x) = \frac{1}{1 + e^{-x}}$ | Output layer for binary classification. |
| **Tanh** | $f(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$ | Hidden layers in RNNs and traditional shallow networks. |
| **ReLU** | $f(x) = \max(0, x)$ | Standard default for hidden layers in CNNs and deep feedforward networks. |
| **Leaky ReLU** | $f(x) = \max(\alpha x, x)$ | Hidden layers where dead neurons occur with standard ReLU. |
| **ELU** | $f(x) = \begin{cases} x & x > 0 \\ \alpha(e^x - 1) & x \le 0 \end{cases}$ | Deep networks requiring smoother negative region learning. |
| **GELU** | $f(x) = x \cdot \Phi(x)$ | Hidden layers in modern Transformer architectures (BERT, GPT). |
| **Swish / SiLU** | $f(x) = \frac{x}{1 + e^{-x}}$ | Deep computer vision backbones (EfficientNet, YOLO). |
| **Softmax** | $f(z)_i = \frac{e^{z_i}}{\sum_{j} e^{z_j}}$ | Output layer for multi-class classification probabilities. |

> loss fuction measures error in predicted output by comparing with real output
5. **Forward Propogation:** The movement from input to prediction is called forward propagation.
6. **BackPropogation :** Calculate error in predicted output. Goes back. Adjust weight. Repeat.
7. **Ephocs**: Learning on all data once is one epoch. More epochs give the model more chances to learn, but too many can cause overfitting.
8. **Batch size :** Training all one million images in one go is not possible. So divide into batchs. Batch size is how many samples the model processes before updating its weights.

## Models

1. **CNN :** (Convolutional Neural Network) Best for images. They learn visual features such as edges, shapes, Objects.

___
2. **RNN :** (Recurrent Neural Network) Designed for sequential data. **It remembers previous input.**   
-> Struggle is storing long sequence data.


___
3. **LSTM :** (Long Short-Term Memory)Improved version of RNN. It remembers important information over longer sequences and forgets irrelevant details.

___
4. **Transformer :** Instead of processing words one by one like an RNN, Transformers look at relationships between all words at once using a mechanism called attention.  
-> **Attention mechanism :** Allows to dynamically focus on the most relevant parts.
