# Supervised learning

___
1. **Linear regression**: Predicts continues values. uses linear equation. y = mx + b

___
2. **Logistic regression** :(Bianry classification or multiclass too) It is classification model despit it's name. Uses sigmoid function.  
-> sigmoid function f = 1/(1+e^-x) -> Converts any value to between 0 to 1.

___
3. **Decision tree :**(Classification) It recursively divides problem. Each node is decision node or leaf node.   
-> Each decision node decides yes or no.  
-> Each leaf node is final classification.  
-> It is like segmentation tree or BST   

___
> Bagging, Boosting  
-> **Bagging** : parallel and independent. Reduce Variance. Resistant to overfitting.  
-> **Boosting** : Sequential and dependent. High accuracy. highly chances of overfitting.  
4. **Random forest :** Multiple decision trees outputs are **paralelly (Bagging)** compared. Final outcomes is **Mode** of all outputs of each decision tree.  
-> High accuracy. Less overfitting
5. **Gradient boost :** (Boosting) Sequential processing. Each tree learns from previous trees mistake.  
-> Very accurate
6. **XGBoost :** Imporvement on gradient boosting.

___
7. **Navie Bayes :**(Classification) Uses probability.

___
8. **KNN(k-nearest neighbours) :** No training. Looks at **k** nearest data points. Then mode of those k points is output of new data point.

___
9. **SVM(Suport vector machine) :** (support both classification and regression)Find best separating boundary in multi demientional space.

## Unsupervised learning

___
> l-bow method to get optimal **k.** Calculate WCSS(Within cluster sum of squares). Plot line graph btw **k and wcss**. Identify Sharp elbow.  

1. **K-Means :** Groups all training data points into k clusters. Calculate centroid of each cluster.  
-> Choose randomly k points. cluster all points to these k points.  
-> Recalculate centroids. Re cluster untill no point changes it'c cluster.  
2. **k-means++ :** Select k points intelligently. First select a point randomly. Then for next point select a point that is at maximum distance. And so on.  

___
3. **DBSCAN and HDBSCAN :** Cluster based on density.Density-based clustering algorithms designed to find clusters of arbitrary shapes and automatically filter out noise (outliers). Unlike K-Means, neither requires specifying the number of clusters (K) upfront.  
> Number of clusters unknown

___
4. **GMM:** Soft clustering. We find probability of a point to be of a cluster.  
> 70% Cluster A  
  30% Cluster B

## Reinforcement learning

___

### Basics
1. **State:** A complete or partial representation of the environment at a specific time step t. The agent decides which action to perform based on current state
2. **Action:** All possible moves or decisions available to the agent within a state. The action is something predefined. 
3. **Environment:** Everything outside the agent — the world in which the agent lives and operates.
4. **Agent:** The learner or decision-maker.
5. **Reward** : Positive feedback. Feedback signals sent from the environment to evaluate the agent's action at time t. 
6. **Penalty:** Negative feedback.

#Imp  

7. **Return**: Cummilative sum of all rewards till time step 't'

___
### Exploration vs exploitation.
> ->**Exploring** every time will lead to lose of reward.  
> ->**Exploiting** (Greedy) may give good reward at the very moment. But may miss  good rewards in future.  
> ->**What should be done When?** Explore till knowledge is full. Exploit after knowledge is satisfiable 

-> Used  $\epsilon$-greedy method.
