# Generative AI
> Traditional ml algorithms will predict output if we give input.  
Gen ai models will return input instead. We give output as input.  
-> Eg: Auto Encoders. They store the data in latent space. By making critical changes to decoder, we can make it generate new data.

1. Generative AI is a branch of AI that creates new content such as text, images, audio, video, or code by learning patterns from large datasets.
2. Types:  
    -> LLM or transformer  
    -> Diffusion model  
    -> GAN(Genarative adversarial networks)  
    -> Variational AutoEncoders


___
* **Transformer**  
-> It used attention mechanism to process word or tocken in context . LLM'S
* **Diffusion Model**   
-> Start with random visual noise. Slowly clean it up to build clear image or video
* **GAN**  
-> It has 2 Neural networks. One is creater and one is judger.
* **Variational autoencoder**  
-> Encode real data. Sligtly make changes to decoder to get completly new data.

____
1. **Prompt.**  
-> The instructions passed to Generative ai models.  
2. **Temperature(0 to 1)**  
-> It controls randomness. Higher the temperature higher the quality of response.
3. **Halucination (IMP)**  
-> Biggest limitations of LLMs. That fabricated but fluent response is called a hallucination.  
-> The model will confidently makeup answer even if it is not there.  
4. **Context window**  
-> Short-term memory. The model need to maintain context. So the summary of the context will be updated with every context.
5. **Embadings**  
-> Represent words in numbers. It is based on NLP. Vector DB.
-> Words with similar meanings end up close together in vector space.
6. **Vector DB**  
-> Stores similar semantic data close to each other in latent space. Low accuracy. High speed. Used in llms.
7. **RAG(Retival augmented generation)**  
-> Instead of completely depending on the training data. _<u>To prevent model from halucination</u>_ 
8. **Fine-tuning**  
-> It is the process of adapting a pre-trained AI model to a specific task or domain by further training it on a smaller, targeted dataset.  
-> It is used for specific task. 
> What is the difference between Fine-Tuning and RAG?  
        Fine-tuning changes the model itself by additional training. RAG leaves the model unchanged and provides it with relevant external information during inference.
